```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
data = {
    "CustomerID": [101, 102, 103, 104, 105,
                   106, 107, 108, 109, 110],

    "CustomerName": ["Anu", "Ravi", "Priya", "John",
                     "Meena", "Arun", "Kavya", "Raj",
                     "Divya", "Kiran"],

    "Branch": ["Bangalore", "Chennai", "Bangalore",
               "Mumbai", "Chennai", "Mumbai",
               "Bangalore", "Delhi", "Delhi", "Chennai"],

    "LoanAmount": [500000, 300000, 700000, 400000,
                   600000, 250000, 800000, 350000,
                   450000, 550000],

    "LoanStatus": ["Approved", "Default", "Approved",
                   "Closed", "Default", "Approved",
                   "Default", "Approved", "Closed", "Approved"]
}

df = pd.DataFrame(data)

print(df)
```

```python
total_loan = df['LoanAmount'].sum()

print("Total Loan Amount:", total_loan)
```

```python
average_loan = df['LoanAmount'].mean()

print("Average Loan Amount:", average_loan)
```

```python
branch_analysis = df.groupby('Branch')['LoanAmount'].sum()

print(branch_analysis)
```

```python
customer_count = df.groupby('Branch')['CustomerID'].count()

print(customer_count)
```

```python
loan_status_count = df['LoanStatus'].value_counts()

print(loan_status_count)
```

```python
branch_analysis = df.groupby('Branch')['LoanAmount'].sum()

branch_analysis.plot(kind='bar')

plt.title('Total Loan Amount by Branch')
plt.xlabel('Branch')
plt.ylabel('Total Loan Amount')
plt.show()
```
